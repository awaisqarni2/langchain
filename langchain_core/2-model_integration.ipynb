# Openai Model Integration

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")

# INIT CHAT MODEL
from langchain.chat_models import init_chat_model
model = init_chat_model("gpt-4.1")

# invoke chat model
response = model.invoke("Hello How are you?")
response
```

```python
# another method for chatmodel
from langchain_openai import ChatOpenAI
model = ChatOpenAI(model_name="gpt-4.1")
reponse = model.invoke("Hellow How are your?")
reponse
```

# Google Gemini Model Integration

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = model.invoke("Why do parrots talk?")
response.content  
```

# Streaming

```python
# Stream (yields chunks progressively)
from langchain_groq import ChatGroq
model = ChatGroq(model="openai/gpt-oss-120b")
for chunk in model.stream("write 200 words about the importance of AI in healthcare"):
    print(chunk.text, end="", flush=True)
```

# Batch

Batching is a collection of independent requests to a model, can significantly improve performance and reduce costs , as the processing can be done in parallel.

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

responses =  model.batch([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
], config={"max_concurrency": 5})  # Max 5 parallel calls

for response in responses:
    print(response)
```
